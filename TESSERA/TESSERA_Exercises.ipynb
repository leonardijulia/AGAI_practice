# GeoTESSERA library

GeoTessera offers two main ways of using the TESSERA embeddings:
1. Stream from the zarr store (using the `GeoTesseraZarr` functionality): read points, regions, and patches straight from the public cloud store, dequantised on their native UTM grid, with nothing downloaded up front.
2. Download tiles (using `GeoTessera`): fetch 0.1-degree tiles to disk and export them as numpy arrays or georeferenced GeoTIFF files, for offline reuse and GIS integration.

```python
gt = GeoTessera()
embedding, crs, transform = gt.fetch_embedding(lon=0.12, lat=52.20, year=2024)

print(f"Shape: {embedding.shape}")  # e.g., (1200, 1200, 128)
print(f"CRS: {crs}")  # UTM projection

# -------

gtz = GeoTesseraZarr()
vec, status = gtz.probe(0.12, 52.20, year=2024) # (128,) float32, 'valid'

print(vec.shape, vec.dtype)
print(status)
```
